## 🔧 Partie 1 : Les Fonctions en Python

Les fonctions sont des blocs de code réutilisables qui effectuent une tâche spécifique. Elles sont essentielles pour écrire du code propre et organisé.

### 1.1 Pourquoi utiliser des fonctions ?

Les fonctions offrent plusieurs avantages :
- **Réutilisabilité** : éviter la répétition de code
- **Lisibilité** : code plus facile à comprendre
- **Maintenance** : modifications centralisées

### 1.2 Définir une fonction simple

```python
# Définition d'une fonction simple sans paramètre
def saluer():
    print("Bonjour, bienvenue dans le monde des fonctions !")

# Pour utiliser la fonction, on l'appelle par son nom suivi de parenthèses
saluer()
```

### 1.3 Fonctions avec paramètres

```python
# Une fonction qui prend un nom en paramètre
def saluer_personne(nom):
    print(f"Bonjour, {nom} ! Ravi de vous rencontrer.")

# On appelle la fonction en lui passant un argument
saluer_personne("Alice")
saluer_personne("Bob")
```

```python
# Une fonction avec plusieurs paramètres
def afficher_infos(nom, age):
    print(f"{nom} a {age} ans.")

afficher_infos("Charles", 25)
afficher_infos("Marie", 30)
```

### 1.4 La valeur de retour (return)

```python
# Une fonction qui additionne deux nombres et retourne le résultat
def additionner(a, b):
    somme = a + b
    return somme

# On stocke le résultat de la fonction dans une variable
resultat_addition = additionner(5, 3)
print("Le résultat de l'addition est :", resultat_addition)

# On peut aussi utiliser la fonction directement
print("La somme de 10 et 20 est :", additionner(10, 20))
```

### 1.5 Fonctions avec valeurs par défaut

```python
# L'âge a une valeur par défaut de 18
def saluer_avec_age(nom, age=18):
    print(f"Salut {nom}, tu as {age} ans.")

# On peut l'appeler sans spécifier l'âge
saluer_avec_age("Martin")

# Ou on peut spécifier une autre valeur pour l'âge
saluer_avec_age("Léa", 25)
```

### 1.6 Exercice pratique : Calculatrice simple

```python
# Créons plusieurs fonctions pour une calculatrice
def multiplier(a, b):
    return a * b

def diviser(a, b):
    if b != 0:
        return a / b
    else:
        return "Erreur : Division par zéro impossible !"

def calculatrice(a, b, operation):
    if operation == "+":
        return additionner(a, b)
    elif operation == "*":
        return multiplier(a, b)
    elif operation == "/":
        return diviser(a, b)
    else:
        return "Opération non supportée"

# Test de notre calculatrice
print("5 + 3 =", calculatrice(5, 3, "+"))
print("5 * 3 =", calculatrice(5, 3, "*"))
print("10 / 2 =", calculatrice(10, 2, "/"))
print("10 / 0 =", calculatrice(10, 0, "/"))
```

## 🏗️ Partie 2 : La Programmation Orientée Objet (POO)

La Programmation Orientée Objet (POO) est une approche de programmation qui permet de modéliser des objets du monde réel à l'aide de classes et d'instances.

### 2.1 Comprendre les Classes et les Objets

- **Une classe** est un modèle ou un plan (comme le plan d'une maison)
- **Un objet** est une instance concrète de cette classe (comme une maison réelle construite à partir du plan)

**Analogie simple :** La classe `Voiture` est le plan. Une voiture rouge de 2022 et une voiture bleue de 2024 sont deux objets différents créés à partir du même plan.

### 2.2 Définir une classe et ses attributs

```python
# On définit la classe Voiture
class Voiture:
    # La méthode __init__ (le constructeur) s'exécute à la création d'un objet.
    # 'self' fait référence à l'objet lui-même.
    def __init__(self, marque, modele, annee):
        # On définit les attributs de l'objet avec les valeurs passées en paramètres.
        self.marque = marque
        self.modele = modele
        self.annee = annee
        self.kilometrage = 0  # Un attribut initialisé par défaut
```

#### Le rôle de `__init__` et `self`

- **`__init__`** : C'est le constructeur. Il est appelé automatiquement pour configurer le nouvel objet.
- **`self`** : C'est une référence à l'instance de la classe en cours de création. C'est grâce à `self` que l'objet peut stocker ses propres attributs.

### 2.3 Créer des objets et accéder aux attributs

```python
# On crée deux objets de la classe Voiture
ma_voiture_rouge = Voiture("Renault", "Clio", 2022)
ma_voiture_bleue = Voiture("Peugeot", "208", 2024)

# On accède aux attributs de chaque objet avec le point "."
print("Ma première voiture :")
print(f"Marque : {ma_voiture_rouge.marque}")
print(f"Modèle : {ma_voiture_rouge.modele}")
print(f"Kilométrage : {ma_voiture_rouge.kilometrage} km")

print("\nMa deuxième voiture :")
print(f"Marque : {ma_voiture_bleue.marque}")
print(f"Modèle : {ma_voiture_bleue.modele}")
```

### 2.4 Les méthodes : les actions de la classe

```python
class Voiture:
    def __init__(self, marque, modele, annee):
        self.marque = marque
        self.modele = modele
        self.annee = annee
        self.kilometrage = 0

    # Méthode pour afficher les informations de la voiture
    def afficher_description(self):
        return f"Marque : {self.marque}, Modèle : {self.modele}, Année : {self.annee}"

    # Méthode pour faire rouler la voiture (elle modifie l'attribut kilometrage)
    def rouler(self, distance):
        self.kilometrage += distance
        print(f"La voiture a parcouru {distance} km. Le nouveau kilométrage est de {self.kilometrage} km.")

# Création d'un objet et appel des méthodes
ma_voiture = Voiture("BMW", "Série 3", 2020)

description = ma_voiture.afficher_description()
print(description)

ma_voiture.rouler(150)
ma_voiture.rouler(50)

# On peut aussi modifier les attributs directement
ma_voiture.annee = 2021
print(f"Année modifiée : {ma_voiture.annee}")
```

### 2.5 L'héritage : Réutiliser du code

L'héritage est un concept clé de la POO. Il permet à une classe d'hériter des attributs et des méthodes d'une autre classe.

- La classe qui hérite est la **classe enfant** (ou dérivée)
- La classe dont on hérite est la **classe parent** (ou de base)

```python
# Classe parent
class Voiture:
    def __init__(self, marque, modele):
        self.marque = marque
        self.modele = modele
        self.est_demarree = False

    def demarrer(self):
        self.est_demarree = True
        print(f"La {self.marque} {self.modele} démarre.")

# Classe enfant : VoitureElectrique hérite de Voiture
class VoitureElectrique(Voiture):
    def __init__(self, marque, modele, autonomie):
        # On appelle le constructeur de la classe parent avec super()
        super().__init__(marque, modele)
        self.autonomie_batterie = autonomie

    def recharger(self):
        print(f"La {self.marque} {self.modele} est en charge.")
        
    def afficher_autonomie(self):
        print(f"Autonomie restante : {self.autonomie_batterie} km")

# Création d'objets
voiture_essence = Voiture("Ford", "Focus")
voiture_electrique = VoitureElectrique("Tesla", "Model 3", 500)

# Les deux objets peuvent utiliser la méthode du parent
voiture_essence.demarrer()
voiture_electrique.demarrer()

# L'objet enfant a ses propres méthodes en plus
voiture_electrique.recharger()
voiture_electrique.afficher_autonomie()
```

### 2.6 Exemple pratique : Système de gestion d'animaux

```python
# Classe de base Animal
class Animal:
    def __init__(self, nom, espece, age):
        self.nom = nom
        self.espece = espece
        self.age = age
        self.energie = 100
    
    def dormir(self):
        self.energie = 100
        print(f"{self.nom} dort et récupère son énergie.")
    
    def manger(self):
        self.energie = min(100, self.energie + 20)
        print(f"{self.nom} mange et gagne de l'énergie.")
    
    def se_presenter(self):
        return f"Je suis {self.nom}, un {self.espece} de {self.age} ans."

# Classe héritée Chat
class Chat(Animal):
    def __init__(self, nom, age, couleur):
        super().__init__(nom, "Chat", age)
        self.couleur = couleur
    
    def miauler(self):
        print(f"{self.nom} fait : Miaou !")
    
    def ronronner(self):
        print(f"{self.nom} ronronne de bonheur.")

# Classe héritée Chien
class Chien(Animal):
    def __init__(self, nom, age, race):
        super().__init__(nom, "Chien", age)
        self.race = race
    
    def aboyer(self):
        print(f"{self.nom} fait : Woof !")
    
    def rapporter(self):
        if self.energie > 20:
            self.energie -= 20
            print(f"{self.nom} rapporte la balle !")
        else:
            print(f"{self.nom} est trop fatigué pour rapporter la balle.")

# Créons nos animaux
mon_chat = Chat("Whiskers", 3, "gris")
mon_chien = Chien("Rex", 5, "Golden Retriever")

# Testons leurs comportements
print(mon_chat.se_presenter())
mon_chat.miauler()
mon_chat.ronronner()

print("\n" + mon_chien.se_presenter())
mon_chien.aboyer()
mon_chien.rapporter()

# Actions communes à tous les animaux
print("\n=== Actions communes ===")
mon_chat.manger()
mon_chien.dormir()
```

### 2.7 Exercice : Compte bancaire simple

```python
class CompteBancaire:
    def __init__(self, proprietaire, solde_initial=0):
        self.proprietaire = proprietaire
        self.solde = solde_initial
        self.historique = []
    
    def deposer(self, montant):
        if montant > 0:
            self.solde += montant
            self.historique.append(f"Dépôt de {montant}€")
            print(f"Dépôt de {montant}€ effectué. Nouveau solde : {self.solde}€")
        else:
            print("Le montant doit être positif.")
    
    def retirer(self, montant):
        if montant > 0 and montant <= self.solde:
            self.solde -= montant
            self.historique.append(f"Retrait de {montant}€")
            print(f"Retrait de {montant}€ effectué. Nouveau solde : {self.solde}€")
        elif montant > self.solde:
            print("Fonds insuffisants !")
        else:
            print("Le montant doit être positif.")
    
    def consulter_solde(self):
        print(f"Solde actuel de {self.proprietaire} : {self.solde}€")
        return self.solde
    
    def afficher_historique(self):
        print(f"\n=== Historique du compte de {self.proprietaire} ===")
        for transaction in self.historique:
            print(f"- {transaction}")
        print(f"Solde final : {self.solde}€\n")

# Test du compte bancaire
compte_alice = CompteBancaire("Alice", 1000)

compte_alice.consulter_solde()
compte_alice.deposer(500)
compte_alice.retirer(200)
compte_alice.retirer(2000)  # Test fonds insuffisants
compte_alice.afficher_historique()
```

## 📋 Résumé et points clés

### Fonctions :
- Utilisez `def` pour définir une fonction
- Les paramètres rendent les fonctions flexibles
- `return` permet de renvoyer des valeurs
- Les valeurs par défaut rendent les paramètres optionnels

### Classes et POO :
- Une **classe** est un modèle (ex: `Voiture`)
- Un **objet** est une instance de cette classe (ex: `ma_voiture`)
- `__init__` est le constructeur qui initialise les attributs
- Les **méthodes** sont les fonctions internes qui agissent sur l'objet
- L'**héritage** permet de créer des classes spécialisées à partir de classes existantes
