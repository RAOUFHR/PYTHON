## Création et Types de Variables

```python
# Variable entière
age = 30
print("Âge :", age, " | Type :", type(age))

# Variable flottante
taille = 1.75
print("Taille :", taille, " | Type :", type(taille))

# Variable chaîne de caractères
prenom = "Alice"
print("Prénom :", prenom, " | Type :", type(prenom))

# Variable booléenne
est_etudiant = True
print("Est étudiant :", est_etudiant, " | Type :", type(est_etudiant))
```

## Manipulation des Variables

```python
# Opérations arithmétiques
a = 10
b = 3
somme = a + b
produit = a * b
division = a / b
reste = a % b

print("Somme :", somme)
print("Produit :", produit)
print("Division :", division)
print("Reste de la division :", reste)

# Comparaisons (le résultat est un booléen)
print("a est-il égal à b ?", a == b)
print("a est-il supérieur à 5 ?", a > 5)
```

##  Les Collections : Listes, Tuples et Dictionnaires

```python
# Création d'une liste de fruits
fruits = ["pomme", "banane", "cerise"]
print("Liste de fruits :", fruits)

# Accéder à un élément (le premier)
premier_fruit = fruits[0]
print("Le premier fruit est :", premier_fruit)

# La taille de la liste avec len()

print("La liste a une taille (longueur) de :", len(fruits))

# Ajouter un élément à la fin
fruits.append("orange")
print("Liste après ajout :", fruits)

# Modifier un élément (le deuxième)
fruits[1] = "mangue"
print("Liste après modification :", fruits)

# Supprimer un élément par sa valeur
fruits.remove("cerise")
print("Liste après suppression :", fruits)

# Supprimer le dernier élément avec pop()
dernier_fruit_supprime = fruits.pop()
print("Le fruit supprimé était :", dernier_fruit_supprime)
print("Liste finale :", fruits)
```

## Une Collection Immuable

```python
# Création d'un tuple de coordonnées
coordonnees = (10, 20)
print("Tuple de coordonnées :", coordonnees)

# On peut toujours accéder aux éléments
x = coordonnees[0]
print("Coordonnée x :", x)

# Cette ligne provoquerait une erreur car on ne peut pas modifier un tuple
# coordonnees[0] = 5
```

## Le Dictionnaire (dict) : Une Collection de Paires Clé-Valeur

```python
# Création d'un dictionnaire pour stocker les informations d'un utilisateur
utilisateur = {
    "nom": "Dupont",
    "prenom": "Jean",
    "age": 25
}

print("Dictionnaire utilisateur :", utilisateur)

# Accéder à la valeur associée à la clé "prenom"
prenom_utilisateur = utilisateur["prenom"]
print("Prénom de l'utilisateur :", prenom_utilisateur)

# Ajouter une nouvelle paire clé-valeur
utilisateur["ville"] = "Paris"
print("Dictionnaire après ajout :", utilisateur)

# Modifier une valeur
utilisateur["age"] = 26
print("Dictionnaire après modification de l'âge :", utilisateur)

# Supprimer une paire clé-valeur
del utilisateur["ville"]
print("Dictionnaire après suppression de la ville :", utilisateur)
```
